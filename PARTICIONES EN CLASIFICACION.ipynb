## GINI

El índice de Gini es una de las formas con las que se optimiza la partición entre nodos de un árbol de decisión. Se encarga de medir y evaluar que tan puros o impuros, es decir que tan uniformes son las clases por cada partición de los árboles. Este valor se calcula con la probabilidad de clasificar incorrectamente en la partición. Se minimiza la función de costo que involucra el índice de Gini de ambos lados del nodo, el cual es un número de 0 a 0.5, y el menor es la partición que se elige para el nodo.

$$ 
\text{Gini} = 1 - \Sigma^k_{i=1} p_i^2
$$

$$
\text{Gini}_{split} = \frac{n_\text{izq}}{n} \text{Gini}_{\text{izq}} + \frac{n_\text{der}}{n} \text{Gini}_{\text{der}}
$$

## ENTROPÍA

La entropía de Shannon, mide la incertidumbre de un sistema, es decir, la cantidad de "sorpresa" que contenga un evento o sistema. Para ello, utiliza la sumatoria de la proporción de cada clase posible, por el logritmo base 2 de la misma proporción, mientras más alta sea la entropía, mayor es el desorden o incertidumbre en la información, y si es 0, indica que hay una sola clase. Se busca que la entropía sea la menor posible, y con ese criterio, se hacen las particiones de los nodos.


$$
\text{Entropía} = - \Sigma^k_{i=1} p_i log_2(p_i)
$$

## LOG-LOSS

El log-loss mide qué tan bien un modelo de clasificación predice la clase correcta. También penaliza mucho si una clase errónea tiene una alta probabilidad asignada. A diferencia de los anteriores, el log loss se aplicaa cada modelo para medir su precisión prediciendo, no como criterio para los nodos.

Fórmula Binaria
$$
\text{LogLoss} = -(y \text{log}(p) + (1-y)\text{log}(1-p))
$$

Fórmula Multiclase
$$
\text{LogLoss} = -\Sigma^k_{i=1} y_i \text{log}(p_i)
$$

## ¿CUÁL ES LA DIFERENCIA ENTRE ENTROPÍA Y LOG-LOSS? 

La principal diferencia es que a pesar de que ambas tienen fórmulas muy similares, la entropía se basa en las proporciones entre clases disponibles, el log loss se basa en las probablidades predichas por el modelo. 

Además, la entropía es un criterio de los nodos, mientras que el log-loss sirve como métrica para un modelo ya entrenado.

## EJEMPLO

```python
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, log_loss
```

```python
# Elegimos el dataset de iris como ejemplo
iris = load_iris()
X = iris.data
y = iris.target
```

```python
#Separamos en Train y Test
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)
```

```python
#Hacemos un modelo de árbol con el criterio de Gini
tree_gini = DecisionTreeClassifier(criterion='gini', random_state=42)
tree_gini.fit(X_train, y_train)
y_pred_gini = tree_gini.predict(X_test)
y_proba_gini = tree_gini.predict_proba(X_test)
```

```python
#Hacemos un modelo de árbol con el criterio de entropía
tree_entropy = DecisionTreeClassifier(criterion='entropy', random_state=42)
tree_entropy.fit(X_train, y_train)
y_pred_entropy = tree_entropy.predict(X_test)
y_proba_entropy = tree_entropy.predict_proba(X_test)
```

```python
# Comparamos cada árbol, con precision y log-loss
print("GINI")
print("Precisión:", accuracy_score(y_test, y_pred_gini))
print("Log Loss:", log_loss(y_test, y_proba_gini))

print("ENTROPÍA")
print("Precisión:", accuracy_score(y_test, y_pred_entropy))
print("Log Loss:", log_loss(y_test, y_proba_entropy))
```

Podemos ver que con un criterio, GINI, la precisión fue más alta, y el log-loss menor, pero con la entropía, la precisión es más baja y el log-loss mayor. Esto quiere decir que para este conjunto de datos, fue más importante la forma en que el índice de GINI mide parte los nodos con las probabilidades de equivocarse en cada uno, y no es tan importante la incertidumbre entre clases que nos la entropía nos ayuda a medir.
